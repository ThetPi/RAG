```python
from dotenv import load_dotenv
import os
import tempfile
from pathlib import Path
from langchain_community.document_loaders import (TextLoader,
    WebBaseLoader,
    DirectoryLoader,
    PyPDFLoader,)

load_dotenv()
```

```python
def load_text_file():
    with tempfile.NamedTemporaryFile(delete=False,suffix='.txt') as temp_file:
        temp_file.write(
            b"Hello, this is a sample text file.\nThis file is used to demonstrate the TextLoader."
        )
        temp_file_path = temp_file.name
    try :
        loader = TextLoader(temp_file_path)
        documents = loader.load()

        for doc in documents:
            print('Doc Content')
            print('doc',doc)
            print('\n\n')
            print('doc.page_content',doc.page_content)
            print('\n\n')
    finally:
        os.remove(temp_file_path)
```

```python
load_text_file()
```

```python
def pdf_loader(pdf_path:str):
    loader = PyPDFLoader(pdf_path)
    documents = loader.load()

    print(f'Loaded number of documents from pdf : {len(documents)}')
    for idx,doc in enumerate(documents):
        print(f'Document {idx+1} .........Preview {doc.page_content[:100]}...')
        print(f'Metadata {doc.metadata}')
        print('\n\n')
        print(10*'*')
        print('\n\n')
```

```python
pdf_loader(pdf_path='doc/langchain_demo.pdf')
```
